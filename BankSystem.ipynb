```python
class Bank:
    user_details={"alextom":{"pas":"at123","bal":1000},"anumol":{"pas":"anu123","bal":2000},"siva":{"pas":"12siva","bal":3000}}
    value=0
    def __init__(self,user,passw):                                             #validating username and password
        if user in self.user_details and passw==self.user_details[user]["pas"]:
            print("Hi, ",user,"How can I help you?")
            print("1. Deposit Cash")
            print("2. Withdraw Cash")
            print("3. Check Balance")
            value=int(input("Please enter the number :"))
            if (value==1):
                self.deposit(user)
            elif (value==2):
                self.withdraw(user)
            elif (value==3):
                self.display_balance(user)
            else:
                print("No operation selected")
        else:
            print("Your USERNAME or PASSWORD is incorrect.")
            
    def deposit(self,user):                                                   #deposit function
        amt=int(input("Enter the amount you want to deposit: "))
        if(amt>0):
            self.user_details[user]["bal"]+=amt
            self.display_balance(user)
        else:
            print("Invalid Input")

    def withdraw(self,user):                                                 #withdraw function
        amt=int(input("Enter the amount you want to withdraw: "))
        if(amt>0 and amt<=self.user_details[user]["bal"]):
            self.user_details[user]["bal"]-=amt
            self.display_balance(user)
        else:
            print("Invalid Input")
            
    def display_balance(self,user):                                         #function for balance display
        print("CURRENT BALANCE :",self.user_details[user]["bal"])
```
