```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("winequality-red.csv",sep=";")
data.head(5)
```

```python
data.describe()
```

```python
data=data.fillna(data.mean())    #1
data.head(5)
```

```python
v_alcohol=data["alcohol"].to_numpy()    #2
print("shape:",v_alcohol.shape)
print(v_alcohol[:5])
```

```python
v_citric=data["citric acid"].to_numpy()
print("shape:",v_citric.shape)
print(v_citric[:5])
```

```python
X=data[["alcohol","citric acid"]].to_numpy()    #3
cov=np.cov(X.T)
print("Covariance Matrix:",cov)
```

```python
evalues,evector= np.linalg.eig(cov)    #4
print("Eigenvalues:",evalues)
print("Eigenvector:",evector)
```

```python
common=data["quality"].mode()      #5
print("Most common wine quality score:",common[0])
print("\nQuality Score Distribution:")
print(data["quality"].value_counts().sort_index())
```
