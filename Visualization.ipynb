```python
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
data=sns.load_dataset("titanic")
print(data.head())
```

1.Countplot: Passenger Count by Embarked Port

```python
plt.figure(figsize=(4, 4))
sns.countplot(x="embark_town",data=data,hue="sex")
plt.title("Countplot: Passenger Count by Embarked Port")
plt.xlabel("Embarked Port")
plt.show()
```

Southampton have the vast majority of passengers, with nearly twice as many men as women.

Cherbourg had a significantly higher proportion of female passengers relative to males compared to Southampton.

Queenstown have the lowest overall embarkation numbers.

2.Swarmplot: Age Distribution across Classes

```python
plt.figure(figsize=(4, 4))
sns.catplot(x="pclass",y="age",kind="swarm",data=data,hue="sex")
plt.title("Swarmplot: Age Distribution across Classes")
plt.xlabel("Class")
plt.show()
```

1st Class passengers with very few children and many passengers aged 40–70+.

2nd and 3rd Classes with a dense concentration of young adults (20–35) and almost all children on board residing in 2nd and 3rd class.

3.Pie Chart: Gender Distribution

```python
plt.figure(figsize=(4, 4))
a=data["sex"].value_counts().reset_index()
print(a)
plt.pie(a["count"],labels=a["sex"],autopct='%.2f%%',explode=[0,0.1],shadow=True)
plt.title("Pie Chart: Gender Distribution")
plt.show()
```

Males made up 64.78% of the passengers on board.

Females comprised 35.24% of total passengers.

4.Heatmap: Correlation Matrix

```python
plt.figure(figsize=(10,8))
col=["age","fare","pclass","sibsp","parch"]
correlation_matrix=data[col].head(30)
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.title("Heatmap: Correlation Matrix")
plt.show()
```

5. Violin Plot: Fare by Class and Gender

```python
plt.figure(figsize=(8,6))
sns.violinplot(data=data,x="pclass",y="fare",hue="sex")
plt.title("Violin Plot: Fare by Class and Gender")
plt.show()
```

1st-class fares have a long right tail extending to extreme outliers. Females in 1st class generally had higher median fares than males.
    
Fares for 2nd and 3rd classes are near the bottom with small variation between men and women.

5.Scatter Plot: Age vs Fare

```python
sns.relplot(data=data,x="age",y="fare",kind="scatter",hue="survived")
plt.title(" Scatter Plot: Age vs Fare")
plt.show()
```

Passengers who paid higher fares were survivors, regardless of age.Age Effect in Low Fares
