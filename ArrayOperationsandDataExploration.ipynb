```python
import numpy as np
import pandas as pd
```

```python
arr1=np.arange(1,11)
print(arr1)
arr2=arr1.reshape(2,5)
print(arr2)
```

```python
arr3=np.arange(1,21)
arr4=arr3[5:15]
print(arr4)
```

```python
print("Mean:",np.mean(arr3))
print("Median:",np.median(arr3))
print("Standard Deviation:",np.std(arr3))
```

```python
arr_x=np.array([[2,3,4,5],[3,4,5,6],[7,6,5,3]])
print("array x:",arr_x)
print("shape of x:",arr_x.shape)
arr_y=np.array([1,2,3,4])
print("\narray y:",arr_y)
print("shape of y",arr_y.shape)
arr_add=arr_x - arr_y
print("\nResult array:")
print(arr_add)
```
